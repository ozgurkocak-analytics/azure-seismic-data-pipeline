```python
from pyspark.sql.functions import col, count, avg, max

# 1. Define paths
storage_name = "stearthquakedata"
silver_path = f"abfss://datalake@{storage_name}.dfs.core.windows.net/2-silver/turkey_quakes_clean"
gold_path = f"abfss://datalake@{storage_name}.dfs.core.windows.net/3-gold/daily_quake_summary"

# 2. Read the Silver Data
df_silver = spark.read.parquet(silver_path)

# 3. Create the Gold Summary (Aggregating by Date and City)
df_gold = df_silver.groupBy("event_date", "city") \
    .agg(
        count("*").alias("total_tremors"),
        avg("magnitude").alias("avg_magnitude"),
        max("magnitude").alias("strongest_tremor")
    ) \
    .orderBy("event_date", col("total_tremors").desc())

# 4. Save as Physical Parquet Files

df_gold.write.mode("overwrite").parquet(gold_path)

print(f"Gold data files updated at: {gold_path}")
```
